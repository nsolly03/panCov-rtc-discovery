```python
import py3Dmol
from pathlib import Path
import numpy as np
from IPython.display import display, Markdown

BASE   = Path('..')
POSES  = BASE / '04-hits/poses'
RECEPT = BASE / '03-virtual-screening'

TARGETS = {
    'NSP12-NSP7': RECEPT / 'NSP12-NSP7_3/receptor_NSP12-NSP7_3.pdb',
    'NSP9-NSP12': RECEPT / 'NSP9-NSP12_5/receptor_NSP9-NSP12_5.pdb',
    'NSP12-NSP8': RECEPT / 'NSP12-NSP8_4/receptor_NSP12-NSP8_4.pdb',
}

LEADS = [
    ('351017',   'NSP12-NSP7', -8.153, 'Scaffold A — Lead #1'),
    ('351017',   'NSP9-NSP12', -8.132, 'Scaffold A — Lead #1'),
    ('351017',   'NSP12-NSP8', -7.759, 'Scaffold A — Lead #1'),
    ('13633807', 'NSP9-NSP12', -8.173, 'Scaffold C — Lead #2'),
    ('5024943',  'NSP12-NSP7', -8.225, 'Scaffold B — Lead #5'),
]
print('Setup OK')
```

```python
def pdbqt_to_pdb_str(pdbqt_path):
    lines, in_m1 = [], False
    with open(pdbqt_path) as f:
        for line in f:
            if line.startswith('MODEL        1'):
                in_m1 = True; continue
            if line.startswith('ENDMDL') and in_m1:
                break
            if in_m1 and line.startswith(('ATOM','HETATM')):
                lines.append(line[:66].rstrip())
    return '\n'.join(lines)

def make_complex(rec_pdb, lig_pdbqt):
    with open(rec_pdb) as f:
        rec = [l.rstrip() for l in f if l.startswith(('ATOM','HETATM','TER'))]
    lig = pdbqt_to_pdb_str(lig_pdbqt).split('\n')
    lig_out = []
    for line in lig:
        if line.startswith(('ATOM','HETATM')):
            line = 'HETATM' + line[6:17] + 'LIG Z   1' + line[26:66]
        lig_out.append(line)
    return '\n'.join(rec + ['TER'] + lig_out + ['END'])

def get_center(pdbqt_path):
    coords, in_m1 = [], False
    with open(pdbqt_path) as f:
        for line in f:
            if line.startswith('MODEL        1'): in_m1 = True
            if line.startswith('ENDMDL') and in_m1: break
            if in_m1 and line.startswith(('ATOM','HETATM')):
                try: coords.append([float(line[30:38]),float(line[38:46]),float(line[46:54])])
                except: pass
    return np.mean(coords, axis=0) if coords else np.zeros(3)

def show_pose(zinc_id, target_name, width=900, height=520):
    lig_path = POSES / target_name / f'{zinc_id}_out.pdbqt'
    if not lig_path.exists():
        print(f'Missing: {lig_path}'); return

    cplx   = make_complex(TARGETS[target_name], lig_path)
    center = get_center(lig_path)

    v = py3Dmol.view(width=width, height=height)
    v.addModel(cplx, 'pdb')

    # Show ALL as cartoon first
    v.setStyle({}, {'cartoon': {'color': 'lightgrey', 'opacity': 0.9}})

    # Override ligand (chain Z, residue LIG) as sticks
    v.setStyle({'resn': 'LIG'},
               {'stick': {'colorscheme': 'greenCarbon', 'radius': 0.25}})

    # Binding pocket surface
    v.addSurface(py3Dmol.SAS,
        {'opacity': 0.12, 'color': 'lightblue'},
        {'resn': {'$ne': 'LIG'}})

    # Ligand centroid
    v.addSphere({'center': {'x': float(center[0]),
                            'y': float(center[1]),
                            'z': float(center[2])},
                 'radius': 1.2, 'color': 'yellow', 'opacity': 0.4})

    v.zoomTo({'resn': 'LIG'})
    v.setBackgroundColor('white')
    v.show()

print('Functions ready')

print('Functions ready')
```

## ZINC351017 — Lead #1 · Scaffold A · Σ = −24.044 kcal/mol

```python
display(Markdown('### NSP9–NSP12 · −8.132 kcal/mol'))
show_pose('351017', 'NSP9-NSP12')
```

```python
display(Markdown('### NSP12–NSP7 · −8.153 kcal/mol'))
show_pose('351017', 'NSP12-NSP7')
```

```python
display(Markdown('### NSP12–NSP8 · −7.759 kcal/mol'))
show_pose('351017', 'NSP12-NSP8')
```

## All Leads Loop

```python
for zinc_id, target, score, label in LEADS:
    display(Markdown(f'### ZINC{zinc_id} · {target} · {score:.3f} kcal/mol · {label}'))
    show_pose(zinc_id, target, width=800, height=450)
```
